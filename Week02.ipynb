```python
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
import matplotlib.pyplot as plt
import numpy as np

img = cv.imread('Foto-KTM.jpg')
cv2_imshow(img) #format BGR

print("Resolusi gambar :",img.shape[1],"x",img.shape[0])

```

```python
img = cv.imread('Foto-KTM.jpg')
img2 = cv.cvtColor(img, cv.COLOR_BGR2RGB) #konversi channel BGR -> RGB
plt.imshow(img2)
```

```python
img = cv.imread('Foto-KTM.jpg')
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB)) #konversi BGR -> RGB digabung
plt.show

print(" -plt.imshow(img2) → digunakan untuk memasukkan gambar ke dalam plot. \n -plt.show() → hanya dipanggil tanpa argumen untuk menampilkan plot.")
```

```python
#menampilkan citra greyscale
img_gray = cv.imread('Foto-KTM.jpg',cv.IMREAD_GRAYSCALE)

#tanpa CMAP
plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title('Tanpa CMAP')

#Dengan CMAP
plt.subplot(1,2,2)
plt.imshow(img_gray,cmap='magma')
plt.title('Dengan CMAP')

plt.show()

```

```python
#Melakukan Resizing
imgRGB = cv.cvtColor(img, cv.COLOR_BGR2RGB)
img_output = cv.resize(imgRGB, (347,229))
plt.imshow(img_output)
```

```python
#Melakukan Flipping
imgFlip = cv.flip(img_output,1)
plt.imshow(imgFlip)
```

```python
#Membuat bentuk Geometri 2D dari OpenCV. Diawali dengan pembuatan black image dengan tipe data int16.
black_img = np.zeros(shape=(530,530,3),dtype=np.int16)
plt.imshow(black_img)

```

```python
#perhatikan koordinat titik2 pt1 & pt2
cv.rectangle(black_img,pt1=(290,0),pt2=(520,130),color=(0,0,255),thickness=10)
plt.imshow(black_img)
```

```python
cv.rectangle(black_img,pt1=(250,250),pt2=(350,350),color=(255,0,0),thickness=10)
plt.imshow(black_img)
```

```python
cv.circle(black_img,center=(150,150),radius=60 ,color=(0,255,0),thickness=8)
plt.imshow(black_img)
```

```python
cv.circle(black_img,center=(400,400),radius=40 ,color=(0,255,0),thickness=-1)
cv.line(black_img,pt1=(0,0),pt2=(530,530),color=(255,255,255),thickness=5)
plt.imshow(black_img)
```

```python
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img,text='PCVK IS FUN',org=(10,500),fontFace=font,fontScale=2,color=(255,255,0),thickness=3,lineType=cv.LINE_AA)
plt.imshow(black_img)
```

```python
black_img2=np.zeros(shape=(530,530,3),dtype=np.int32)
plt.imshow(black_img2)
```

```python
vertices = np.array([[150,350],[250,250],[450,350],[250,450]],dtype=np.int32)
pts = vertices.reshape((-1,1,2))
cv.polylines(black_img2,[pts],isClosed=True,color=(0,255,0),thickness=5)
plt.imshow(black_img2)
```
